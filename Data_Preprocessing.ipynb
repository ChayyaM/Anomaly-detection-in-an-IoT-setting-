### Library 

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

### Data Loading and Preprocessing

```python
# Load Data
df = pd.read_csv("environmental_design_dataset.csv")

# Parse timestamp
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Drop irrelevant columns
columns_to_drop = ["Sensor_ID", "Zone", "CO2", "PM25", "Illuminance", "PMV","PPD","DA", "AAI",
                   "NDVI", "Comfort_Score", "Illuminance", "AQI_Class"]

df = df.drop(columns=columns_to_drop)

# Handling missing values
df = df.dropna()

# Remove duplicates
df = df.drop_duplicates()

# Select the columns of sensor features
sensor_features = ["Temp", "Humidity", "Sound"]
print(df)
```

### Saving the cleaned and preprocessed dataset

```python
df.to_csv("cleaned_sensor_data.csv", index=False)
```
